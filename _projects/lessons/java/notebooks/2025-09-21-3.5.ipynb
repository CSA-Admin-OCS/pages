## 1. Reference Guide

### Key Topics

| Term | Definition | Example |
| --- | --- | --- |
| **Method** | A named behavior defined inside a class. | `calculateArea(8, 5)` |
| **Return type** | The type sent to the caller; `void` means no value. | `int`, `boolean`, `void` |
| **Parameter** | A typed input declared in a method header. | `int length` |
| **Argument** | A value supplied when calling a method. | `8` in `calculateArea(8, 5)` |
| **Return statement** | Ends the method and can send a value to the caller. | `return length * width;` |
| **Method signature** | For these methods, the name and ordered parameter types. | `calculateArea(int, int)` |
| **Accessor** | Returns a copy of a field's value; has a non-void return type. | `getLength()` |
| **Mutator** | Updates a field; often has a void return type. | `setLength(10)` |

### Reading a Method Header

<pre><code class="language-java">public int calculateArea(int length, int width)</code></pre>

| Part | Purpose | Example |
| --- | --- | --- |
| Access modifier | Controls where the method can be called | `public` |
| Return type | States what comes back to the caller | `int` |
| Method name | Describes the behavior | `calculateArea` |
| Parameters | Declare the typed inputs, in order | `int length, int width` |

The header introduces the method; the braces enclose its **body**. The return type and parameter names are not part of its signature. See [Oracle's Defining Methods guide](https://docs.oracle.com/javase/tutorial/java/javaOO/methods.html).

### Choosing a Method's Behavior

- Use `void` for an action such as printing or changing a field. A `void` method can use `return;` to exit early, but cannot return a value.
- A non-void call supplies one value. Java evaluates the expression in `return`, then gives its value to the caller (**return by value**). Its type must fit the declared return type: `return length * width;` fits an `int` method; `return "40";` does not. Printing does not supply a return value.
- Every path that finishes a value-returning method must return a compatible value. A `return` inside an `if` or loop exits the entire method. Execution resumes at the call site; later statements in that method are skipped. An unconditional statement directly after `return` is unreachable and causes a compile error.
- Arguments match parameters by position and compatible type. Primitive arguments are copied: changing an `int` parameter does not change the caller's variable.

This lesson focuses on instance methods with primitive inputs and results (Topic 3.5). Object-reference parameters and return values belong to Topic 3.6; class methods are covered in Topic 3.7. `main` is provided only to run each example.

### A. `void` Methods and Returning Methods

`printDimensions` performs an action and sends no value back. `calculateArea` returns an `int` that the caller stores and then prints. Predict two output lines: `Dimensions: 8 by 5` and `Area: 40`.

```java
// CODE_RUNNER: Run both methods, change the dimensions, and explain which call supplies a value to the caller.
public class Shape {
    // This method performs an action and returns no value.
    public void printDimensions(int length, int width) {
        System.out.println("Dimensions: " + length + " by " + width);
    }

    // This method returns an int that the caller can store.
    public int calculateArea(int length, int width) {
        return length * width;
    }

    public static void main(String[] args) {
        Shape rectangle = new Shape();
        rectangle.printDimensions(8, 5);

        int area = rectangle.calculateArea(8, 5);
        System.out.println("Area: " + area);
    }
}
Shape.main(null);
```

### B. Parameters, Argument Order, and Primitive Copies

Java matches arguments to parameters from left to right. Swapping `8` and `5` changes the description even though their product would be the same.

In `doubleLength`, the parameter receives a copy of the caller's `length`. Reassigning that parameter changes the copy. Predict the final two lines: the returned value is `16`, while the caller's variable is still `8`.

```java
// CODE_RUNNER: Swap the arguments, then explain why doubleLength returns 16 while the caller still holds 8.
public class ArgumentOrder {
    public void printDimensions(int length, int width) {
        System.out.println("length=" + length + ", width=" + width);
    }

    public int doubleLength(int length) {
        length = length * 2;
        return length;
    }

    public static void main(String[] args) {
        ArgumentOrder example = new ArgumentOrder();
        example.printDimensions(8, 5);
        example.printDimensions(5, 8);
        int length = 8;
        int doubled = example.doubleLength(length);
        System.out.println("Returned: " + doubled);
        System.out.println("Caller length: " + length);
    }
}
ArgumentOrder.main(null);
```

### C. Return Paths in Branches and Loops

`isValidSquare` returns a primitive `boolean` on all three paths. A nonpositive dimension returns `false` immediately; equal positive dimensions return `true`; otherwise it returns `false`.

Predict the three outputs: `true`, `false`, `false`. Try `(0, 0)` and explain why the first guard takes priority over equality. Removing the last `return` would leave a path without a result and cause a compile error.

`firstSquareSide` checks whole-number sides from 1 through 10 and returns the first side whose square meets `minimumArea`. The input is a positive integer. Predict `firstSquareSide(9)` → `3` and `firstSquareSide(101)` → `-1`.

When side `3` meets the target, `return side;` exits both the loop and the method: it skips sides 4–10 and the final `return -1;`. If all ten sides fail, the loop finishes and the final return runs. Unlike `break`, `return` does not continue with statements after the loop. Try targets `1`, `10`, and `100`.

**Trace check:** For target `9`, record the sides tested, the returned value, and why `-1` is skipped. For target `101`, explain why `-1` is reached. Add these two traces to your submission notes.

```java
// CODE_RUNNER: Trace returns from a branch and a loop; test targets 9 and 101 and explain whether the final return runs.
public class RectangleCheck {
    public boolean isValidSquare(int length, int width) {
        if (length <= 0 || width <= 0) {
            return false;
        }
        if (length == width) {
            return true;
        }
        return false;
    }

    public int firstSquareSide(int minimumArea) {
        for (int side = 1; side <= 10; side++) {
            if (side * side >= minimumArea) {
                return side;
            }
        }
        return -1;
    }

    public static void main(String[] args) {
        RectangleCheck checker = new RectangleCheck();
        System.out.println(checker.isValidSquare(4, 4));
        System.out.println(checker.isValidSquare(8, 5));
        System.out.println(checker.isValidSquare(-1, 5));
        System.out.println("First side for 9: " + checker.firstSquareSide(9));
        System.out.println("First side for 101: " + checker.firstSquareSide(101));
    }
}
RectangleCheck.main(null);
```

### D. Accessors and Mutators

An accessor returns a copy of a field value so callers can read it. A mutator updates that field. Here `getLength` returns an `int`, while `setLength` returns nothing and updates the object. Mutators are often `void`, but a method that updates a field and returns a result is still a mutator.

Predict `8` before the call and `10` after it. `savedLength` keeps the earlier copy, so the last line is still `8`. Unlike changing a primitive parameter in Example B, assigning the object's field persists after the method returns. A caller in another class can use these public methods without directly accessing the private field. For this example, use positive lengths.

```java
// CODE_RUNNER: Predict the field and saved accessor result after setLength; explain why the saved copy stays 8.
public class RectangleState {
    private int length = 8;

    public int getLength() {
        return length;
    }

    public void setLength(int newLength) {
        length = newLength;
    }

    public static void main(String[] args) {
        RectangleState rectangle = new RectangleState();
        int savedLength = rectangle.getLength();
        System.out.println("Before: " + savedLength);
        rectangle.setLength(10);
        System.out.println("After: " + rectangle.getLength());
        System.out.println("Saved copy: " + savedLength);
    }
}
RectangleState.main(null);
```
